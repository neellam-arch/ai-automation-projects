# Project 1: Student Enquiry Sorter

**Problem:** Student enquiries arrive mixed together, and some are missing contact details.

**What it does:** sorts enquiries by course, flags missing emails, and saves one JSON file per course.

**Skills:** Python, JSON, filtering, data-quality checks, file output

## 1. Sample data
In a real automation this would come from a web form, email or Google Sheet.

```python
import json

enquiries = [
  {"name": "Ali", "course": "PTE UKVI", "email": "ali@x.com"},
  {"name": "Hina", "course": "Python", "email": "hina@x.com"},
  {"name": "Omar", "course": "PTE UKVI"},
  {"name": "Zara", "course": "AI Tools", "email": "zara@x.com"},
  {"name": "Bilal", "course": "Python"}
]
```

## 2. Filter students by course
Works like a Filter node in n8n.

```python
def get_students_by_course(data, course_name):
    result = []
    for student in data:
        if student["course"] == course_name:
            result.append(student)
    return result
```

## 3. Data-quality check: missing emails

```python
def find_missing_emails(data):
    missing = []
    for student in data:
        if not student.get("email"):
            missing.append(student["name"])
    return missing
```

## 4. Save one file per course, then report

```python
courses = ["PTE UKVI", "Python", "AI Tools"]

for course in courses:
    students = get_students_by_course(enquiries, course)
    filename = course.replace(" ", "_").lower() + ".json"
    with open(filename, "w") as f:
        json.dump(students, f, indent=2)
    print(f"Saved {len(students)} students to {filename}")

print("Missing emails:", find_missing_emails(enquiries))
```
